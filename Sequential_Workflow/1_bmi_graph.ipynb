```python
from langgraph.graph import StateGraph, START, END
from typing import TypedDict
```

```python
# define state
class BMIState(TypedDict):
    height: float
    weight: float
    answer: float
    catagory : str
```

```python
def calculator(state: BMIState) -> BMIState:
    height = state["height"]
    weight = state["weight"]
    bmi = weight / (height ** 2)
    state["answer"] = round(bmi, 2)
    return state

def label(state: BMIState) -> BMIState:
    bmi = state["answer"]
    if bmi < 18.5:
        state["catagory"] = "Underweight"
    elif 18.5 <= bmi < 25:
        state["catagory"] = "Normal weight"
    elif 25 <= bmi < 30:
        state["catagory"] = "Overweight"
    else:
        state["catagory"] = "Obesity"
    return state
```

```python
# define graph
graph = StateGraph(BMIState)
# add nodes
graph.add_node("calculator", calculator)
graph.add_node("label", label)
# add edges
graph.add_edge(START, "calculator")
graph.add_edge("calculator", "label")
graph.add_edge("label", END)
# compile graph
workflow = graph.compile()
```

```python
intial_state = {
    "height": 1.75,
    "weight": 70.0}
final_state = workflow.invoke(intial_state)

print(final_state["answer"], final_state["catagory"])
```
