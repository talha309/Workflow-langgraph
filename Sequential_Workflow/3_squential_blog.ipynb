```python
from langgraph.graph import StateGraph, START, END
from langchain_google_genai import ChatGoogleGenerativeAI
from typing import TypedDict
from dotenv import load_dotenv
import os
```

```python
load_dotenv()  # Load environment variables from .env file
GOOGLE_API_KEY = os.getenv("GOOGLE_API_KEY")  # Get the Google API
```

```python
llm = ChatGoogleGenerativeAI(model="gemini-3.5-flash", temperature=0.7, api_key=GOOGLE_API_KEY)
```

```python
class BlogState(TypedDict):
    topic: str
    outline: str
    content: str
```

```python
def create_outline(state: BlogState) -> BlogState:
    try:
        topic = state["topic"]
        prompt = f"Create a detailed outline for a blog post about {topic}."
        outline = llm.invoke(prompt)
        state["outline"] = outline
        return state
    except Exception as e:
        print(f"Error in create_outline: {e}")
        raise

def create_content(state: BlogState) -> BlogState:
    try:
        outline = state["outline"]
        prompt = f"Write a blog post based on the following outline:\n{outline}"
        content = llm.invoke(prompt)
        state["content"] = content
        return state
    except Exception as e:
        print(f"Error in create_content: {e}")
        raise
```

```python
graph = StateGraph(BlogState)
# add nodes
graph.add_node("Create_Outline", create_outline)
graph.add_node("Create_Content", create_content)
# add edges
graph.add_edge(START, "Create_Outline")
graph.add_edge("Create_Outline", "Create_Content")
graph.add_edge("Create_Content", END)
# compile the graph
workflow = graph.compile()
```

```python
initial_state = {
    "topic": "Technical SEO"
}
final_state = workflow.invoke(initial_state)
outline = final_state["outline"]
content = final_state["content"]
print( outline.content[0]["text"])
```
